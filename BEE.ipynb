1. Fragment Length Frequency + Rescaling

```python
import pandas as pd
import matplotlib.pyplot as plt
import gzip

# Load data from .bed.gz
def load_bed_data(filepath):
    data = []
    with gzip.open(filepath, 'rt') as f:
        for line in f:
            parts = line.strip().split()
            start = int(parts[1])
            end = int(parts[2])
            length = end - start
            data.append(length)
    return data

fragment_lengths = load_bed_data('query.bed.gz')

# Normalize and plot
import numpy as np
from collections import Counter

counts = Counter(fragment_lengths)
total = sum(counts.values())
normalized_freq = {k: v/total for k, v in counts.items()}

# Plot
plt.figure(figsize=(10, 5))
plt.plot(list(normalized_freq.keys()), list(normalized_freq.values()), marker='o')
plt.xlabel('Fragment Length (bp)')
plt.ylabel('Normalized Frequency [A.U.]')
plt.title('Fragment Length Distribution')
plt.grid(True)
plt.show()
```

3. Markov Transition Matrix

```python
sequence = "AGCTTAGCTAGGCTA"

# Initialize matrix
states = ['A', 'C', 'G', 'T']
matrix = {s1: {s2:0 for s2 in states} for s1 in states}

# Count transitions
for i in range(len(sequence) - 1):
    matrix[sequence[i]][sequence[i+1]] += 1

# Normalize
for s1 in matrix:
    total = sum(matrix[s1].values())
    if total > 0:
        for s2 in matrix[s1]:
            matrix[s1][s2] /= total

import pandas as pd
df = pd.DataFrame(matrix).T
print("Markov Transition Matrix:")
display(df)
```

4. Fasta: Multiline to Single-line Converter

```python
def convert_fasta(input_file, output_file):
    with open(input_file) as fin, open(output_file, 'w') as fout:
        seq = ''
        for line in fin:
            line = line.strip()
            if line.startswith(">"):
                if seq:
                    fout.write(seq + '\n')
                    seq = ''
                fout.write(line + '\n')
            else:
                seq += line
        if seq:
            fout.write(seq + '\n')

convert_fasta("multiline_input.fasta", "converted.fasta")
```
